```python
date_df= pd.read_csv('../../data_dump/Comments.csv')
```

```python
date_df
```

```python
i=0
while(i < date_df.shape[0]):
    if pd.isnull(date_df.Link[i]):
        date_df.Message[i-1]= str(date_df.Message[i-1]) + str(date_df.Date[i])
        date_df.Message[i-1]= date_df.Message[i-1].replace('nan', '')
        date_df.drop(date_df.index[i], inplace= True)
        date_df.reset_index(drop= True, inplace= True)
        i = i - 1
    i = i + 1
```

```python
date_df.head()
```

```python
date_df.shape
```

```python
date_df.drop_duplicates(subset= 'Link', inplace= True)
```

```python
date_df.shape
```

```python
post_df= pd.merge(post_df, date_df, how= 'left', left_on= 'post_link', right_on= 'Link')
```

```python
likes_df.like_names= likes_df.like_names.fillna("['NaN']").apply(literal_eval)
```

```python
like_names = [item for sublist in likes_df.like_names.tolist() for item in sublist]
```

```python
likes_counts= pd.DataFrame({'names': like_names})
likes_counts= likes_counts[likes_counts.names!= 'NaN']
```

```python
likes_counts.names= likes_counts.names.apply(lambda x: x.split('\n')[1].strip())
```

```python
likes_counts.to_excel('../likes.xlsx', index= False)
```

```python
likes_df.comments_replies_names= likes_df.comments_replies_names.fillna("['NaN']").apply(literal_eval)
```

```python
replies_names = [item for sublist in likes_df.comments_replies_names.tolist() for item in sublist]
```

```python
replies_count= pd.DataFrame({'names': replies_names})
replies_count= replies_count[replies_count.names!= 'NA']
```

```python
replies_count.to_excel('../replies.xlsx', index= False)
```
