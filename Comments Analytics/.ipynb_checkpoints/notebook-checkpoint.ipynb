```python
def return_likes(df):
    raw_names= list(itertools.chain.from_iterable(df.like_names.tolist()))
    headlines= list(itertools.chain.from_iterable(df.like_headlines.tolist()))
    links= list(itertools.chain.from_iterable(df.like_profile_links.tolist()))
    likes_df= pd.DataFrame({'name': raw_names, 'headlines': headlines, 'links': links})
    likes_df= likes_df[likes_df.name != 'NaN']
    return likes_df
```

```python
likes_counts= return_likes(likes_df)
```

```python
def replies_df(df):
    df.comments_replies_content= df.comments_replies_content.fillna("['NaN']")
    df.comments_replies_names= df.comments_replies_names.fillna("['NaN']")
    df.comments_replies_headlines= df.comments_replies_headlines.fillna("['NaN']")
    df.comments_replies_links= df.comments_replies_links.fillna("['NaN']")
    df['comments_replies_content']= df['comments_replies_content'].apply(literal_eval)
    df['comments_replies_names']= df['comments_replies_names'].apply(literal_eval)
    df['comments_replies_headlines']= df['comments_replies_headlines'].apply(literal_eval)
    df['comments_replies_links']= df['comments_replies_links'].apply(literal_eval)
    names= list(itertools.chain.from_iterable(df.comments_replies_names.tolist()))
    headlines= list(itertools.chain.from_iterable(df.comments_replies_headlines.tolist()))
    links= list(itertools.chain.from_iterable(df.like_profile_links.tolist()))
    likes_df= pd.DataFrame({'name': names, 'headlines': headlines, 'links': links})
    likes_df= likes_df[likes_df.name != 'NaN']
    return likes_df
```

```python
replies_df(likes_df)
```

```python
import pandas as pd
dump_df= pd.read_csv("../data_dump/Comments.csv")
```

```python
dump_df.columns
```

```python
dump_df.Link.isna().sum()
```

```python
pd.set_option('display.max_rows', 800)
```

```python
i=0
while(i < dump_df.shape[0]):
    if pd.isnull(dump_df.Link[i]):
        dump_df.Message[i-1]= str(dump_df.Message[i-1]) + str(dump_df.Date[i])
        dump_df.Message[i-1]= dump_df.Message[i-1].replace('nan', '')
        dump_df.drop(dump_df.index[i], inplace= True)
        dump_df.reset_index(drop= True, inplace= True)
        i= i-1
    i = i + 1
```
