```python
import re
def striphtml(data):
    p = re.compile(r'<.*?>')
    return p.sub('', data)
```

```python
df.comments_content= df.comments_content.apply(striphtml)
```

```python
def clean(data):
    return re.sub(r"(Likes)|(Like)", "", data)
df.Comments_likes= df.Comments_likes.apply(clean)
```

```python
def clean(data):
    return re.sub(r"(No)", "0", data)
df.Comments_likes= df.Comments_likes.apply(clean)
```

```python
df.Comments_likes= df.Comments_likes.astype('int64')
```

```python
def clean(data):
    return re.sub(r"(Replies)|(Reply)", "", data)
df.Comments_replies= df.Comments_replies.apply(clean)
```

```python
def clean(data):
    return re.sub(r"(No)", "0", data)
df.Comments_replies= df.Comments_replies.apply(clean)
```

```python
df.Comments_replies=  df.Comments_replies.astype('int64')
```

```python
import spacy
```

```python
nlp= spacy.load('en_core_web_lg')
```

```python
doc= nlp(df.comments_content[1])
```

```python
df['NERs']= ""
for i,con in enumerate(df.comments_content):
    doc= nlp(con)
    df.NERs[i]= doc.ents
```

```python
pd.set_option('display.max_rows', 300)
```

```python
df['PROPN']= ""
for i,con in enumerate(df.comments_content):
    doc= nlp(con)
    tokens= list()
    for token in doc:
        if token.pos_ == 'PROPN':
            tokens.append(token)
    df['PROPN'][i] = tokens
```

```python
df.PROPN= [str(p) for p in df.PROPN]
```

```python
def get_str(text):
    return re.findall('\[(.*?)\]', text)[0]
df.PROPN= df.PROPN.apply(get_str)
```

```python
df.PROPN
```

```python
df['Unknown']= ""
for i,con in enumerate(df.comments_content):
    doc= nlp(con)
    tokens= list()
    for token in doc:
        if token.pos_ == 'X':
            tokens.append(token)
    df['Unknown'][i] = tokens
```

```python
df= df.drop(['NERs', 'Unknown', 'NOUN'], axis= 1)
```

```python
words_list= []
for w in df.PROPN:
    try:
        if ',' in w:
            words_list.extend(w.split(','))
        else:
            words_list.append(w)
    except Exception:
        pass
```

```python
words_list= [word.lower() for word in words_list]
```

```python
words_list= [word.strip() for word in words_list]
```

```python
words= pd.DataFrame({'word': words_list})
```

```python
keys= list(words.word.value_counts().index[:11])
```

```python
values= list(words.word.value_counts().values[:11])
```

```python
del keys[3]
del values[3]
```

```python
pd.Series(values, index= keys)
```

```python
sns.set_style('white')
def show_values_on_bars(axs):
    def _show_on_single_plot(ax):        
        for p in ax.patches:
            _x = p.get_x() + p.get_width() / 2
            _y = p.get_y() + p.get_height() + 0.03
            value = '{:.0f}'.format(p.get_height())
            ax.text(_x, _y, value, ha="center", fontsize=14) 

    if isinstance(axs, np.ndarray):
        for idx, ax in np.ndenumerate(axs):
            _show_on_single_plot(ax)
    else:
        _show_on_single_plot(axs)
fig, ax = plt.subplots(figsize= (12, 6))
g= sns.barplot(x= [w.upper() for w in keys],
            y= values, color="lightblue")
plt.setp(ax.patches, linewidth=0)
plt.title("MOST FREQUENT APP NAMES", fontsize= 14)
plt.xticks(fontsize=14, rotation=45)
plt.yticks(fontsize=14);
show_values_on_bars(ax)
plt.show()
```
