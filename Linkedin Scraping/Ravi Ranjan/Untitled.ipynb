```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(df.text[112], 'html.parser')

print(soup.prettify())
```

```python
import re
def striphtml(data):
    p = re.compile(r'<.*?>')
    return p.sub('', data)
```

```python
df.text= df.text.apply(striphtml)
```

```python
df.text= df.text.str.strip()
```

```python
df= df.iloc[:113]
```

```python
df.to_csv('text_cleaned.csv', index= False)
```
