```python
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
pd.set_option('display.max_columns', 100) # Show 100 columns when looking at dataframe
import itertools
import re
```

```python
import plotly.offline as pyo
import plotly.graph_objs as go
pyo.init_notebook_mode(connected=True)
```

```python
df= pd.read_csv('Likes_Data.csv')
```

```python
df.columns
```

```python
lst= df.Names.value_counts()
```

```python
del lst['Shreya Mukerjee']
```

```python
del lst['Avijit Nandy']
```

```python
data = [go.Bar(
    x=lst[:5].index.to_list(),  
    y=lst[:5].to_list(),
    width= [0.5]*5
)]
layout = go.Layout(
    title= go.layout.Title(text="My Ardent Followers Based on Likes",xref="paper",x=0),
    font=dict(family='Arial', size= 14),
    xaxis_tickfont_size=13
    )
fig = go.Figure(data=data, layout=layout)
fig.update_traces(marker_color='lightblue', marker_line_color='black',
                  marker_line_width=1.0)
fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=False)
fig.update_xaxes(tickangle=340, tickfont=dict(family='Rockwell', color='black', size=14))
pyo.plot(fig)
```
