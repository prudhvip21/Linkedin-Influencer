```python
import pandas as pd
import re
```

```python
df= pd.read_csv('top_voices.csv')
```

```python
df.date= df.date.apply(lambda x: x.split('\n  \n')[:-1])
```

```python
def clean_date(text):
    text= [re.sub(r'( • )', '', t) for t in text]
    text= [t.strip() for t in text]
    text= [re.sub(r'(Edited)?', '', t) for t in text]
    text= [re.sub(r'(Published)?', '', t) for t in text]
    return text
```

```python
df.date= df.date.apply(clean_date)
```

```python
test_list= ['7mo', '8mo', '9mo', '10mo', '11mo', '12mo', '1yr']
```

```python
def rem_dat(text):
    return [x for x in text if x not in test_list]
```

```python
df.date= df.date.apply(rem_dat)
```

```python
import ast
```

```python
df.likes= df.likes.apply(ast.literal_eval)
```

```python
df.columns
```

```python
df.comments= df.comments.apply(ast.literal_eval)
```

```python
df.columns
```

```python
df.followers= df.followers.apply(lambda x: int(x.replace(',', '')))
```

```python
df.likes= df.likes.apply(lambda x: [int(t.replace(',', '')) for t in x])
```

```python
df.comments= df.comments.apply(lambda x: [t.replace(' Comments', '') for t in x])
```

```python
df.comments= df.comments.apply(lambda x: [t.replace(' Comment', '') for t in x])
```

```python
df.comments= df.comments.apply(lambda x: [int(t.replace(',', '')) for t in x])
```

```python
df.head()
```

```python
for i in range(df.shape[0]):
    df.likes[i]= df.likes[i][:len(df.date[i])]
    df.comments[i]= df.comments[i][:len(df.date[i])]
```

```python
df['avg_likes']= ''
df['avg_comments']= ''
import numpy as np
for i in range(df.shape[0]):
    df.avg_likes[i]= np.mean(df.likes[i])
    df.avg_comments[i]= np.mean(df.comments[i])
```

```python
df.head()
```

```python
import numpy as np
for i in range(df.shape[0]):
    df.num_posts[i]= len(df.date[i])
```

```python
df.to_csv('clean_data.csv', index= False)
```

```python
df['likes_per_thousand']= ''
for i in range(df.shape[0]):
    df.likes_per_thousand[i]= [like/df.followers[i] * 1000 for like in df.likes[i]]
```

```python
df.likes_per_thousand
```

```python
df['commnets_per_thousand']= ''
for i in range(df.shape[0]):
    df.commnets_per_thousand[i]= [comment/df.followers[i] * 1000 for comment in df.comments[i]]
```

```python
df['avg_likes_per_thousand']= ''
df['avg_comments_per_thousand']= ''
import numpy as np
for i in range(df.shape[0]):
    df.avg_likes_per_thousand[i]= np.mean(df.likes_per_thousand[i])
    df.avg_comments_per_thousand[i]= np.mean(df.commnets_per_thousand[i])
```

```python
df.to_csv('new_clean_data.csv', index= False)
```
